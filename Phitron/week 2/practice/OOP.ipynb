```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    def __init__(self, vtype, name, speed):
        self.__type = vtype
        self.__name = name
        self.__speed = speed

    def type_getter(self):
        return self.__type

    def name_getter(self):
        return self.__name

    def speed_getter(self):
        return self.__speed

    def classify(self):
        if self.__speed > 100:
            return "High-speed Vehicle"
        else:
            return "Normal Vehicle"

    @abstractmethod
    def show_Info(self):
        pass


class Car(Vehicle):
    def __init__(self, vtype, name, speed):
        super().__init__(vtype, name, speed)

    def show_Info(self):
       
        print(f"Type: {self.type_getter()}, Name: {self.name_getter()}, Speed: {self.speed_getter()} km/h, Predicted: {self.classify()}")


class Bike(Vehicle):
    def __init__(self, vtype, name, speed):
        super().__init__(vtype, name, speed)

    def show_Info(self):
        
        print(f"Type: {self.type_getter()}, Name: {self.name_getter()}, Speed: {self.speed_getter()} km/h, Predicted: {self.classify()}")




t = int(input("Number of Vehicles? "))
vehicles = []

while t>0:
    data = input("Enter vehicle info (Type Name Speed): ").split()
    cls_name, name, speed = data[0], data[1], int(data[2])

    if cls_name == "Car":
        obj = Car(cls_name, name, speed)
    elif cls_name == "Bike":
        obj = Bike(cls_name, name, speed)
    else:
        print("Invalid Vehicle Type!")
        continue

    print(f"{cls_name} Added: {name}")
    vehicles.append(obj)
    t-=1

print("--Vehicle Details--")
# Display all info
for v in vehicles:
    v.show_Info()
```
