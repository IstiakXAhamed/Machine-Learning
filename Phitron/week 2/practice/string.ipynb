Create a text file named emails.txt containing multiple email addresses, some in uppercase and some repeated.
Write a Python program to clean the data by converting all emails to lowercase, removing duplicates, sorting them alphabetically, and saving the cleaned list to a new file named unique_emails.txt.

```python
with open("emails.txt",'r') as file:
    emailList = file.readlines()
    emailLower = list(map(lambda x: x.lower() , emailList))
    print(len(emailLower))
    emailSet = set(emailLower)
    uniqueMails = list(emailSet)
    uniqueMails.sort()
    #uniqueMails = list(map(lambda x: x.strip() , uniqueMails))
    print(len(uniqueMails))
    print(uniqueMails)

with open("unique_email.txt",'w') as file :
    mailsList = file.writelines(uniqueMails)    
```

Create a text file named names.txt containing a list of names.
Some names may have extra spaces, inconsistent capitalization (like “aLiCe” or “ bob ”), or appear more than once.
Write a Python program to clean the data by:
1.  Removing extra spaces.
2.  Converting all names so that only the first letter is capitalized.
3.  Removing duplicate names. 
4. Sorting the names alphabetically.
Finally, save the cleaned list into a new file named clean_names.txt.


```python
with open("names.txt",'r+') as file:
    content= file.readlines()
    print(content)
    content = list(map(lambda x: x.strip(),content))
    print(content)
    content = list(map(lambda x: " ".join(x.split()),content))
    print(content)
    content = list(map(lambda x: x.title(),content))
    print(content)
    print(len(content))
    content = list(set(content))
    print(len(content))
    content.sort()
    print(content)
content= list(map(lambda x : x+ "\n",content))

with open("clean_names.txt",'w') as file :
    clean_names = file.writelines(content)
    
    
```

Problem:
 Given two sets of friends from two people, find the mutual friends, unique friends of each, and the total number of unique friends.

Example Input:
a_friends = {"Rahim", "Karim", "Sakib", "Jamal"}
b_friends = {"Sakib", "Jamal", "Rafiq", "Nadim"}

Expected Output:
Mutual friends: {'Sakib', 'Jamal'}
Unique to A: {'Rahim', 'Karim'}
Unique to B: {'Rafiq', 'Nadim'}
Total unique friends: 6


```python
a_friends = {"Rahim", "Karim", "Sakib", "Jamal"}
b_friends = {"Sakib", "Jamal", "Rafiq", "Nadim"}

print(f"Mutual friends : {a_friends.intersection(b_friends)}")
print(f"Unique to A : {a_friends.difference(b_friends)}")
print(f"Unique to B : {b_friends.difference(a_friends)}")
print(f"Total Unique friends: {len(a_friends.union(b_friends))}")
```

Problem 2: Sort Students by Marks (Tuple + Dictionary)
Problem:
 You are given a list of tuples — each containing a student’s name and mark.
 Sort them in descending order of marks and print the top 3 scorers.
Example Input:
students = [("Rafi", 89), ("Sumi", 95), ("Hasan", 90), ("Nila", 75), ("Anik", 98)]

Expected Output:
Top 3 students:
Anik - 98
Sumi - 95
Hasan - 90


```python
students = [("Rafi", 89), ("Sumi", 95), ("Hasan", 90), ("Nila", 75), ("Anik", 98)]
students.sort(key=lambda x:x[1],reverse=True)
print(students)
print("Top 3 Students :")
print(f"{students[0][0]} - {students[0][1]} ")
print(f"{students[1][0]} - {students[1][1]} ")
print(f"{students[2][0]} - {students[2][1]} ")
```

Problem 3: Word Length Filter (Lambda + Filter)
Problem:
 Given a list of words, use filter() and lambda to return only words whose length is greater than 4.
 Example Input:
words = ["sun", "planet", "moon", "star", "universe"]

Expected Output:
['planet', 'universe']


```python
words = ["sun", "planet", "moon", "star", "universe"]
word_filter = list(filter(lambda x: len(x)>4,words))
print(word_filter)
```

Problem 4: Square of Even Numbers (Map + Filter + Lambda)
Problem:
 Write a one-line Python expression using map(), filter(), and lambda that takes a list of integers and returns the squares of even numbers only.

```python
nums = [1, 2, 3, 4, 5, 6]
even_squares = list(map(lambda x: x**2, filter(lambda x: x % 2 == 0, nums)))
print(even_squares)
```
