```python
import pandas as pd
import numpy as np
import seaborn as sns

# Pandas Zero to Advanced Tutorial

# 1. Basic Setup
import matplotlib.pyplot as plt

# Display settings for better visualization
pd.set_option('display.max_columns', None)
pd.set_option('display.expand_frame_repr', False)
plt.style.use('seaborn-whitegrid')

# 2. Introduction to Pandas Objects
# Creating a Series
s = pd.Series([1, 3, 5, np.nan, 6, 8])
print("Basic Series:")
print(s)

# Creating a DataFrame from a dictionary
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 24, 35, 32],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [50000, 60000, 70000, 80000]
}
df = pd.DataFrame(data)
print("\nBasic DataFrame:")
print(df)

# 3. Reading and Writing Data
# Uncomment these to try with actual files
# df_csv = pd.read_csv('example.csv')
# df_excel = pd.read_excel('example.xlsx')
# df_sql = pd.read_sql('SELECT * FROM table', connection)

# Writing data
# df.to_csv('output.csv', index=False)
# df.to_excel('output.xlsx', index=False)
# df.to_json('output.json')

# 4. Data Inspection
print("\nData Inspection:")
print("Shape:", df.shape)
print("\nInfo:")
print(df.info())
print("\nDescription:")
print(df.describe())
print("\nFirst 2 rows:")
print(df.head(2))
print("\nLast 2 rows:")
print(df.tail(2))

# 5. Selection and Indexing
print("\nSelection and Indexing:")
print("Select column 'Name':")
print(df['Name'])
print("\nSelect multiple columns:")
print(df[['Name', 'Age']])
print("\niLoc (by position):")
print(df.iloc[0:2, 0:2])
print("\nLoc (by label):")
print(df.loc[0:1, ['Name', 'Age']])

# 6. Data Cleaning
# Create a DataFrame with some missing values
df_missing = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8],
    'C': [9, 10, 11, np.nan]
})
print("\nDataFrame with missing values:")
print(df_missing)

# Check for missing values
print("\nMissing values:")
print(df_missing.isnull().sum())

# Fill missing values
print("\nFill missing values with 0:")
print(df_missing.fillna(0))

# Drop rows with missing values
print("\nDrop rows with missing values:")
print(df_missing.dropna())

# 7. Data Manipulation
# Add a new column
df['Experience'] = [3, 5, 10, 7]
print("\nAdd a new column:")
print(df)

# Apply a function
df['Bonus'] = df['Salary'].apply(lambda x: x * 0.1)
print("\nApply a function to create a new column:")
print(df)

# 8. Grouping and Aggregation
print("\nGrouping and Aggregation:")
# Create a larger DataFrame for demonstration
df_larger = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value1': [10, 20, 30, 40, 50, 60],
    'Value2': [100, 200, 300, 400, 500, 600]
})
print(df_larger)

# Group by Category and compute the mean
print("\nGroup by Category and compute the mean:")
print(df_larger.groupby('Category').mean())

# Multiple aggregations
print("\nMultiple aggregations:")
print(df_larger.groupby('Category').agg({
    'Value1': ['min', 'max', 'mean'],
    'Value2': ['sum', 'mean']
}))

# 9. Merging, Joining, and Concatenating
# Create DataFrames for demonstration
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['John', 'Anna', 'Peter', 'Linda']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 5],
    'Age': [28, 24, 35, 40]
})

print("\nMerging DataFrames:")
print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)

# Inner merge
print("\nInner merge:")
print(pd.merge(df1, df2, on='ID', how='inner'))

# Outer merge
print("\nOuter merge:")
print(pd.merge(df1, df2, on='ID', how='outer'))

# Concatenation
print("\nConcatenation:")
print(pd.concat([df1, df1], ignore_index=True))

# 10. Time Series with Pandas
print("\nTime Series with Pandas:")
# Create a date range
dates = pd.date_range('20230101', periods=6)
print("Date range:")
print(dates)

# Create a DataFrame with dates as index
df_time = pd.DataFrame(np.random.randn(6, 4), index=dates, columns=list('ABCD'))
print("\nDataFrame with dates as index:")
print(df_time)

# Resample to monthly frequency
print("\nResample to monthly frequency:")
print(df_time.resample('M').mean())

# 11. Visualization
print("\nVisualization examples:")
# Uncomment to display plots
# df['Age'].plot(kind='hist', title='Age Distribution')
# plt.show()

# df.plot(x='Age', y='Salary', kind='scatter', title='Age vs Salary')
# plt.show()

# sns.heatmap(df.corr(), annot=True)
# plt.title('Correlation Matrix')
# plt.show()

# 12. Advanced Operations
print("\nAdvanced Operations:")
# Pivot tables
df_pivot = pd.DataFrame({
    'Date': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02'],
    'Product': ['A', 'B', 'A', 'B'],
    'Sales': [100, 200, 150, 250]
})
print("Original DataFrame for pivoting:")
print(df_pivot)

pivot = df_pivot.pivot_table(index='Date', columns='Product', values='Sales')
print("\nPivot table:")
print(pivot)

# Cross-tabulation
print("\nCross-tabulation:")
print(pd.crosstab(df_pivot['Date'], df_pivot['Product']))

# 13. Working with Text Data
print("\nWorking with Text Data:")
# Create a DataFrame with text data
df_text = pd.DataFrame({
    'Text': ['Python is great', 'Pandas is powerful', 'Data science is fun', 'Machine learning is exciting']
})
print(df_text)

# String methods
print("\nContains 'Python':")
print(df_text['Text'].str.contains('Python'))

print("\nSplit by space:")
print(df_text['Text'].str.split(' ', expand=True))

# 14. Customizing Pandas
print("\nCustomizing Pandas:")
# Set display options
pd.set_option('display.max_rows', 10)
pd.set_option('display.precision', 2)

# 15. Conclusion
print("\nThis completes the zero to advanced Pandas tutorial!")
```
