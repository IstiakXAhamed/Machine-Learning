```python

arr = np.array([10, 20, 30, 40, 50])
arr[1:4]   # [20 30 40]
arr[:3]    # [10 20 30]
arr[3:]    # [40 50]
arr[::2]   # [10 30 50] (step=2)(prori 2 step e ekta kore nichi , like 0 te 1 ta then 2nd e 1 ta )

```

```python
arr = np.arange(1,10).reshape(3,3)
print(arr)
print(arr[:,[1]])
print()
print(arr[:,1:2])
print(arr[arr>5])
print(arr[1:,])
```

```python

arr = np.array([10, 20, 30, 40, 50])
new_arr = np.where(arr > 25, 1, 0)
print(new_arr)  # Output: [0 0 1 1 1]
print()
arr = np.array([10, 20, 30, 40, 50])
indexes = np.argwhere(arr > 25)
print(indexes)  # Output: [[2], [3], [4]]
print()

arr = np.array([10, 20, 30, 40])
sample = np.random.choice(arr, size=(2,4))
sample = np.random.choice(arr, size=2)
print(sample)  # Example: [30 10]
```
